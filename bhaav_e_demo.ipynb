# Bhaav-E: text to animated sticker

A 212M-parameter text-to-video model, trained from scratch, that turns a prompt into a 2-second looping sticker
(256 × 256, 8 fps).

Prompts work best in the training style: *subject, motion, effect, background colour*. The model knows 30 subjects
(emotion faces, animated emoji, UI animations), 12 effects and 12 backgrounds; the demo's prompt builder lists them.
